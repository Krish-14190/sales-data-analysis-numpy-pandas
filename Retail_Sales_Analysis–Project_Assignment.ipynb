## Part 3: Sales Analysis 

```python
df
```

```python
# Calculate total sales, total profit and total quantity sold.
print("Total Sales " ,df['Sales'].sum())
print("Total Quantity " ,df['Quantity'].sum())
print("Total Profit ",df['Profit'].sum())
```

```python
# Find the average sales, average profit and average discount.
print(f"Average Sales : {df['Sales'].mean()}")
print(f"Average Profit : {df['Profit'].mean()}")
print(f"Average Discount : {df['Discount'].mean()}")
```

```python
# Find the Top 5 products based on total sales.
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)
```

```python
#  Find the Top 10 customers based on total sales.
df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
```

```python
# Calculate category-wise total sales.
df.groupby('Category')['Sales'].sum()
```

```python
# Calculate region-wise total profit.
df.groupby('Region')['Profit'].sum()
```

```python
# Find the highest and lowest sales orders.
print("Highest Sale ",df.sort_values(by="Sales",ascending=False).head(1)[['Product Name','Order ID','Sales']])
print("Lowest Sale",df.sort_values(by="Sales",ascending=True).head(1)[['Product Name','Order ID','Sales']])
```

## Part 4: Advanced Pandas

```python
df
```

```python
# Find the Top 5 most profitable products.
df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(5)
```

```python
# Identify all loss-making products.
loss_products = df.groupby('Product Name')['Profit'].sum()
print(loss_products[loss_products<0].sort_values())
```

```python
df.groupby('Product Name')['Profit'].sum().min()
```

```python
# Calculate category-wise and sub-category-wise profit using groupby().
df.groupby(['Category','Sub-Category'])['Profit'].sum()
```

```python
# Create a pivot table showing total sales for each region and category.
df.pivot_table(values='Sales',index='Region',columns='Category',aggfunc=sum)
```

```python
# Calculate monthly sales for the entire dataset.
df['Order Months'] = df['Order Date'].dt.month
print(df.groupby('Order Months')['Sales'].sum().sort_index())
```

```python
# Find which month generated the highest sales.
df.groupby('Order Months')['Sales'].sum().sort_values(ascending=False).head(1)
```

```python
# Calculate the average order value (AOV)
Total_Sum = df['Sales'].sum()
Unique_Orders = df['Order ID'].nunique()
AOV = Total_Sum/Unique_Orders
print("AOV : ", AOV)
```

## NumPy – Retail Sales Data Analysis

```python
# Calculate the average sales using NumPy.
np.mean(df['Sales'])
```

```python
# Calculate the median sales of all orders.
np.median(df['Sales'])
```

```python
# Calculate the standard deviation of sales.
np.std(df['Sales'])
```

```python
# Find the minimum and maximum sales values.
print("Maximum sale value : ", np.max(df['Sales']))
print("Minimun sale value : ", np.min(df['Sales']))
```

```python
# Classify each order as High Sales or Low Sales based on average sales.
avg_sales = np.mean(df['Sales'])

df['Sales Category'] = np.where(df['Sales'] >= avg_sales, 'High Sales', 'Low Sales')

print(df[['Sales', 'Sales Category']])
```

```python
# Count the total number of loss-making transactions (Profit less than zero).
loss_count = np.sum(df['Profit'] < 0)

print("Total loss-making transactions:", loss_count)
```
